# Phase 4: Exploratory Data Analysis (EDA)
## Academic Risk Analysis

This notebook performs comprehensive Exploratory Data Analysis on the cleaned student dataset (`data/processed/cleaned_final_marks.csv`).

### Objectives:
1. **Univariate Analysis**: Examine summary statistics, distributions, and skewness.
2. **Outlier Analysis**: Detect and quantify potential outliers using IQR.
3. **Bivariate Analysis**: Analyze correlations and key feature relationships with Final Exam Marks.
4. **Academic Risk Grouping**: Segment students into Risk Levels and evaluate subgroup profiles.
5. **Export Visualizations**: Save generated figures to `output/figures/`.

--- 
### 1. Import Libraries & Set Styles

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Styling
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update({'font.sans-serif': 'DejaVu Sans', 'figure.max_open_warning': 0})

# Ensure output directories exist
os.makedirs("../output/figures", exist_ok=True)
```

--- 
### 2. Load Cleaned Dataset

```python
df = pd.read_csv("../data/processed/cleaned_final_marks.csv")
print(f"Dataset Shape: {df.shape}")df.head()
```

--- 
### 3. Univariate Analysis (Summary Statistics & Distributions)

```python
num_cols = [
    "Attendance (%)",
    "Internal Test 1 (out of 40)",
    "Internal Test 2 (out of 40)",
    "Assignment Score (out of 10)",
    "Daily Study Hours",
    "Final Exam Marks (out of 100)"
]

summary_stats = df[num_cols].describe().T
summary_stats["median"] = df[num_cols].median()
summary_stats["skewness"] = df[num_cols].skew()
summary_stats
```

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle("Univariate Feature Distributions", fontsize=16, fontweight="bold")

for idx, col in enumerate(num_cols):
    ax = axes[idx // 3, idx % 3]
    sns.histplot(df[col], kde=True, ax=ax, color="#2b5c8f", bins=20)
    ax.set_title(f"Distribution of {col}", fontsize=12, fontweight="bold")

plt.tight_layout()
plt.show()
```

--- 
### 4. Outlier Analysis

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle("Outlier Detection via Boxplots", fontsize=16, fontweight="bold")

for idx, col in enumerate(num_cols):
    ax = axes[idx // 3, idx % 3]
    sns.boxplot(y=df[col], ax=ax, color="#e67e22")
    ax.set_title(f"Boxplot of {col}", fontsize=12, fontweight="bold")

plt.tight_layout()
plt.show()
```

--- 
### 5. Correlation & Relationship Analysis

```python
plt.figure(figsize=(10, 8))
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, linewidths=0.5)
plt.title("Correlation Matrix Heatmap", fontsize=14, fontweight="bold")
plt.show()
```

--- 
### 6. Risk Group Analysis

```python
def categorize_risk(marks):
    if marks < 50:
        return "High Risk (<50)"
    elif marks < 75:
        return "Moderate (50-74)"
    else:
        return "High Performer (>=75)"

df["Academic_Risk_Group"] = df["Final Exam Marks (out of 100)"].apply(categorize_risk)
df.groupby("Academic_Risk_Group")[num_cols].mean().round(2)
```
